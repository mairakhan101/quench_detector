```python
import os
import numpy as np
import matplotlib.pyplot as plt


# ----------------------------
# Configuration
# ----------------------------
DATA_DIR_CANDIDATES = [
    ".",             # current folder
    "/mnt/data",      # where your uploaded files live in this environment
]

FILES = {
    "6p_norm": "coilparam_MQXFA_test_6p_norm.dat",
    "6p_skew": "coilparam_MQXFA_test_6p_skew.dat",
    "8p_norm": "coilparam_MQXFA_test_8p_norm.dat",
    "8p_skew": "coilparam_MQXFA_test_8p_skew.dat",
}

# MATLAB:
# pointsZi(1,1)=0.0+0.075*i;
# pointsZi(2,1)=0.0+0.076*i;
pointsZi = np.array([0.0 + 0.075j, 0.0 + 0.076j], dtype=np.complex128)

# MATLAB:
measRadius = 0.048383
orientAngle = 0.0  # radians


# ----------------------------
# Helpers: locate and read coilparam data
# ----------------------------
def find_data_dir() -> str:
    for d in DATA_DIR_CANDIDATES:
        ok = True
        for fn in FILES.values():
            if not os.path.exists(os.path.join(d, fn)):
                ok = False
                break
        if ok:
            return d
    raise FileNotFoundError(
        "Could not find all coilparam files. Searched:\n" + "\n".join(DATA_DIR_CANDIDATES)
    )



def read_coilparam(path: str):
    """
    Parses files like:
      length    nturns    R      Theta
      VERTEX RBUCK  1.0000000000e+00  1.00  4.8500000000e-02  7.4176493210e-01
    Returns:
      lengths, nturns, R, theta, header_line
    """
    lengths, nturns, R, theta = [], [], [], []
    header = None

    with open(path, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            if header is None:
                header = line
                continue

            parts = line.split()
            if len(parts) < 6:
                continue
            if parts[0].upper() != "VERTEX":
                continue

            try:
                L = float(parts[2])
                N = float(parts[3])
                r = float(parts[4])
                th = float(parts[5])
            except ValueError:
                continue

            lengths.append(L)
            nturns.append(N)
            R.append(r)
            theta.append(th)

    if len(R) == 0:
        raise ValueError(f"No coil vertices parsed from file: {path}")

    return (
        np.array(lengths, dtype=float),
        np.array(nturns, dtype=float),
        np.array(R, dtype=float),
        np.array(theta, dtype=float),
        header,
    )


# ----------------------------
# MATLAB functions translated (CALCULATION MUST STAY IN FILE ORDER)
# ----------------------------
def fluxCalc(QAvertZ: np.ndarray, pointsZi_: np.ndarray) -> np.ndarray:
    """
    MATLAB:
      Avp(jjvert)=Avp(jjvert)+(-1)^jji*-1*log(zEval(jjvert)-pointsZi(jji));

    Using enumerate(start=1) to match MATLAB's 1-based exponent.
    """
    QAvertZ = np.asarray(QAvertZ, dtype=np.complex128).reshape(-1)
    pointsZi_ = np.asarray(pointsZi_, dtype=np.complex128).reshape(-1)

    Avp = np.zeros(QAvertZ.shape, dtype=np.complex128)
    for j, p in enumerate(pointsZi_, start=1):
        Avp += ((-1) ** j) * (-1) * np.log(QAvertZ - p)
    return Avp


def alternating_signs(n: int) -> np.ndarray:
    """
    MATLAB alt_pm1:
      [+1, -1, +1, -1, ...]
    IMPORTANT: This must align with FILE ORDER, not sorted theta order.
    """
    s = np.ones(n, dtype=float)
    s[1::2] *= -1.0
    return s


def plotCoilparamVert(R: np.ndarray, theta: np.ndarray, titleText: str):
    """
    MATLAB-like visualization of the vertices paired in file order (segment connections).
    This is a geometry/ordering plot, not a 'field vs theta' plot.
    """
    x = R * np.cos(theta)
    y = R * np.sin(theta)

    plt.figure()
    for i in range(0, len(R) - 1, 2):
        plt.plot(x[i:i+2], y[i:i+2], "-o")
    plt.plot(x[0::2], y[0::2], "rs", label="odd index (MATLAB 1,3,5...)")
    plt.plot(x[1::2], y[1::2], "ks", label="even index (MATLAB 2,4,6...)")
    plt.title(titleText.replace("_", r"\_"))
    plt.axis("equal")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.show()


import numpy as np

def A_complex_potential(Z, pointsZi):
    """Same field model as fluxCalc, but vectorized for arbitrary Z grid/points."""
    A = np.zeros_like(Z, dtype=np.complex128)
    for j, p in enumerate(pointsZi, start=1):         # MATLAB indexing
        A += ((-1)**j) * (-1) * np.log(Z - p)         # same sign convention
    return A

def qa_response_from_disturbance(R, theta, pointsZi, *,
                                z0=0.0+0.0j, rotation=0.0,
                                use_alt_signs=True):
    """
    Compute the QA winding response for a given winding geometry and placement.

    Inputs:
      R, theta: coil vertices from coilparam file (file order)
      pointsZi: disturbance source points (complex)
      z0: antenna center offset (complex), i.e. x0 + 1j*y0
      rotation: antenna rotation (rad)
      use_alt_signs: apply +1,-1,+1,-1,... like MATLAB

    Returns:
      QAvertZ_placed: placed vertices (complex)
      Avp: A(z) at each vertex
      signs: +/- pattern
      contrib: Avp*signs
      total: sum(contrib) = winding output
    """
    # Base coil vertices in complex plane (coil geometry)
    QAvertZ = R * np.exp(1j * theta)

    # Apply placement: rotate then translate
    QAvertZ_placed = z0 + np.exp(1j * rotation) * QAvertZ

    # Evaluate disturbance "potential" at placed vertices
    Avp = A_complex_potential(QAvertZ_placed, pointsZi)

    if use_alt_signs:
        signs = np.ones(len(Avp))
        signs[1::2] *= -1
    else:
        signs = np.ones(len(Avp))

    contrib = Avp * signs
    total = np.sum(contrib)

    return QAvertZ_placed, Avp, signs, contrib, total



# ----------------------------
# Physically meaningful plotting utilities
# ----------------------------
def plot_sources_and_vertices(QAvertZ: np.ndarray, label: str):
    plt.figure()
    plt.scatter(np.real(QAvertZ), np.imag(QAvertZ), s=20, label="coil vertices")
    plt.scatter(np.real(pointsZi), np.imag(pointsZi), s=60, marker="s", label="sources (pointsZi)")
    plt.title(f"Complex plane: coil vertices + sources ({label})")
    plt.axis("equal")
    plt.xlabel("Re(z)")
    plt.ylabel("Im(z)")
    plt.legend()
    plt.show()

def field_vector_map(pointsZi, xlim=(-0.06, 0.06), ylim=(-0.01, 0.11),
                     n=300, arrow_step=10, label="disturbance field"):
    """
    Physically oriented visualization of the field derived from the complex potential A(z).

    A(z) = Σ s_j log(z - p_j)   (same sign convention as fluxCalc)
    dA/dz = Σ s_j / (z - p_j)

    Interpret φ = Re(A) as scalar potential -> E = -∇φ:
        Ex = -Re(dA/dz)
        Ey =  Im(dA/dz)

    Plots:
      1) |E| magnitude heatmap (what you'd call "field strength" in real life)
      2) streamlines of E (field lines)
      3) quiver arrows (direction + relative strength)
    """
    x = np.linspace(xlim[0], xlim[1], n)
    y = np.linspace(ylim[0], ylim[1], n)
    X, Y = np.meshgrid(x, y)
    Z = X + 1j * Y

    # Same sign convention as fluxCalc (MATLAB indexing starts at 1)
    s_list = []
    for j in range(1, len(pointsZi) + 1):
        s_list.append(((-1) ** j) * (-1))  # matches your fluxCalc multiplier

    # Compute dA/dz = Σ s_j / (z - p_j)
    dAdz = np.zeros_like(Z, dtype=np.complex128)
    for s, p in zip(s_list, pointsZi):
        dAdz += s / (Z - p)

    # Field from scalar potential φ=Re(A): Ex=-Re(dA/dz), Ey=Im(dA/dz)
    Ex = -np.real(dAdz)
    Ey =  np.imag(dAdz)
    Emag = np.sqrt(Ex**2 + Ey**2)

    # Mask near singularities to avoid blow-up in plots
    mask = np.zeros_like(Z, dtype=bool)
    for p in pointsZi:
        mask |= np.abs(Z - p) < 5e-4
    Exm = np.ma.array(Ex, mask=mask)
    Eym = np.ma.array(Ey, mask=mask)
    Em  = np.ma.array(Emag, mask=mask)

    # 1) Magnitude map
    plt.figure(figsize=(7, 5))
    plt.contourf(X, Y, Em, levels=60)
    plt.colorbar(label="|E| (field magnitude, arbitrary units)")
    plt.scatter(np.real(pointsZi), np.imag(pointsZi), marker="s", s=80, label="source points")
    plt.title(f"Field magnitude |E| ({label})")
    plt.xlabel("x (m)")
    plt.ylabel("y (m)")
    plt.axis("equal")
    plt.legend()
    plt.show()

    # 2) Streamlines (field lines)
    plt.figure(figsize=(7, 5))
    plt.contourf(X, Y, Em, levels=60)
    plt.colorbar(label="|E| (arbitrary units)")
    plt.streamplot(X, Y, Exm.filled(0.0), Eym.filled(0.0), density=1.4)
    plt.scatter(np.real(pointsZi), np.imag(pointsZi), marker="s", s=80, label="source points")
    plt.title(f"Field lines (streamlines of E) over |E| ({label})")
    plt.xlabel("x (m)")
    plt.ylabel("y (m)")
    plt.axis("equal")
    plt.legend()
    plt.show()

    # 3) Quiver arrows (downsampled)
    plt.figure(figsize=(7, 5))
    plt.contourf(X, Y, Em, levels=60)
    plt.colorbar(label="|E| (arbitrary units)")

    xs = X[::arrow_step, ::arrow_step]
    ys = Y[::arrow_step, ::arrow_step]
    exs = Exm[::arrow_step, ::arrow_step].filled(0.0)
    eys = Eym[::arrow_step, ::arrow_step].filled(0.0)

    plt.quiver(xs, ys, exs, eys, angles="xy")
    plt.scatter(np.real(pointsZi), np.imag(pointsZi), marker="s", s=80, label="source points")
    plt.title(f"Vector field arrows (E) over |E| ({label})")
    plt.xlabel("x (m)")
    plt.ylabel("y (m)")
    plt.axis("equal")
    plt.legend()
    plt.show()




def plot_vs_index(y: np.ndarray, label: str, ylabel: str, title: str):
    """
    Index plots are always safe: file order is the natural traversal used in the sum.
    """
    plt.figure()
    plt.plot(np.arange(len(y)), y, "-o")
    plt.title(f"{title} ({label})")
    plt.xlabel("vertex index (file order)")
    plt.ylabel(ylabel)
    plt.show()


def plot_vs_theta_sorted(theta: np.ndarray, y: np.ndarray, label: str, ylabel: str, title: str):
    """
    Sort ONLY for visualization. Do not feed sorted arrays back into physics sums.
    """
    idx = np.argsort(theta)
    th = theta[idx]
    yy = y[idx]

    plt.figure()
    plt.plot(th, yy, "-o")
    plt.title(f"{title} (theta-sorted) ({label})")
    plt.xlabel("theta (rad)")
    plt.ylabel(ylabel)
    plt.show()


def plot_scatter_theta(theta: np.ndarray, y: np.ndarray, label: str, ylabel: str, title: str):
    """
    Scatter avoids implying continuity when theta order jumps around.
    """
    plt.figure()
    plt.scatter(theta, y, s=25)
    plt.title(f"{title} (scatter, file order) ({label})")
    plt.xlabel("theta (rad)")
    plt.ylabel(ylabel)
    plt.show()


def plot_sign_pattern(signs: np.ndarray, label: str):
    plt.figure()
    plt.step(np.arange(len(signs)), signs, where="mid")
    plt.ylim(-1.5, 1.5)
    plt.title(f"Alternating sign pattern (+1,-1,...) used in sum ({label})")
    plt.xlabel("vertex index (file order)")
    plt.ylabel("sign")
    plt.show()


# ----------------------------
# One winding run (MATLAB-correct calculation, physically meaningful plots)
# ----------------------------
def run_winding(dat_path: str, label: str):
    """
    This matches MATLAB numerically:
    - Read data in file order
    - Apply orientAngle to theta (still in file order)
    - Compute QAvertZ, Avp in file order
    - Apply alternating signs in file order
    - Sum in file order

    For plots:
    - Geometry plot in file order (paired segments)
    - Avp and contrib shown vs index (file order)
    - Also shown vs theta with sorting (for physical interpretation)
    """
    L, N, R, theta, header = read_coilparam(dat_path)
    theta = theta - orientAngle  # MATLAB: QA.data(:,4)=QA.data(:,4)-orientAngle;

    # 1) Geometry / coil layout
    plotCoilparamVert(R, theta, f"{label} coil vertices (paired segments, file order)")

    # 2) Complex vertex positions
    QAvertZ = R * np.exp(1j * theta)
    plot_sources_and_vertices(QAvertZ, label)

    # 3) Flux calc (file order)
    Avp = fluxCalc(QAvertZ, pointsZi)

    # Plots of Avp:
    # - vs index (what you are traversing in the loop)
    plot_vs_index(np.real(Avp), label, "Re(Avp)", "Re(Avp) vs vertex index")
    plot_vs_index(np.imag(Avp), label, "Im(Avp)", "Im(Avp) vs vertex index")

    # - vs theta (physically meaningful view). Use sorted theta to avoid zig-zags.
    plot_vs_theta_sorted(theta, np.real(Avp), label, "Re(Avp)", "Re(Avp) vs theta")
    plot_vs_theta_sorted(theta, np.imag(Avp), label, "Im(Avp)", "Im(Avp) vs theta")

    # Optional: scatter in file order (also avoids criss-cross)
    plot_scatter_theta(theta, np.real(Avp), label, "Re(Avp)", "Re(Avp) vs theta")
    plot_scatter_theta(theta, np.imag(Avp), label, "Im(Avp)", "Im(Avp) vs theta")

    # 4) Alternating signs (file order)
    signs = alternating_signs(len(Avp))
    plot_sign_pattern(signs, label)

    contrib = Avp * signs

    # Signed contributions:
    plot_vs_index(np.real(contrib), label, "Re(Avp * sign)", "Signed contribution vs vertex index")
    plot_vs_theta_sorted(theta, np.real(contrib), label, "Re(Avp * sign)", "Signed contribution vs theta")

    # 5) Total flux (complex sum, file order)
    total = np.sum(contrib)

    # 6) MATLAB dlmwrite equivalent (KEEP FILE ORDER like MATLAB)
    out = np.column_stack([theta, np.real(contrib)])
    out_name = "flux_" + os.path.basename(dat_path)
    np.savetxt(out_name, out, delimiter=" ", fmt="%.10e")
    print(f"Wrote {out_name}")

    return total


# ----------------------------
# Main execution
# ----------------------------
def main():
    data_dir = find_data_dir()
    print("Using data directory:", data_dir)

    # Run all four (same as MATLAB blocks)
    norm6p = run_winding(os.path.join(data_dir, FILES["6p_norm"]), "6p normal")
    skew6p = run_winding(os.path.join(data_dir, FILES["6p_skew"]), "6p skew")
    norm8p = run_winding(os.path.join(data_dir, FILES["8p_norm"]), "8p normal")
    skew8p = run_winding(os.path.join(data_dir, FILES["8p_skew"]), "8p skew")

    # MATLAB:
    # C6p=real(norm6p)+i*real(skew6p)
    # C8p=real(norm8p)+i*real(skew8p)
    C6p = np.real(norm6p) + 1j * np.real(skew6p)
    C8p = np.real(norm8p) + 1j * np.real(skew8p)

    # MATLAB:
    # qloc=4/3*C6p/C8p*measRadius
    qloc = (4.0 / 3.0) * (C6p / C8p) * measRadius
    # Visualize the disturbance field in space (independent of coil geometry)
    field_vector_map(pointsZi, label="from log sources (same as fluxCalc)")


    print("\n--- Results (MATLAB-equivalent) ---")
    print("norm6p =", norm6p)
    print("skew6p =", skew6p)
    print("norm8p =", norm8p)
    print("skew8p =", skew8p)
    print("C6p   =", C6p)
    print("C8p   =", C8p)
    print("qloc  =", qloc)

    # Physically interpretable summary plots:
    plt.figure()
    plt.plot([0, 1], [np.abs(C6p), np.abs(C8p)], "-o")
    plt.xticks([0, 1], ["|C6p|", "|C8p|"])
    plt.ylabel("magnitude")
    plt.title("Harmonic magnitudes")
    plt.show()

    plt.figure()
    plt.plot([0, 1], [np.angle(C6p), np.angle(C8p)], "-o")
    plt.xticks([0, 1], ["∠C6p", "∠C8p"])
    plt.ylabel("phase (rad)")
    plt.title("Harmonic phases")
    plt.show()

    plt.figure()
    plt.plot([0, 1], [np.real(qloc), np.imag(qloc)], "-o")
    plt.xticks([0, 1], ["Re(qloc)", "Im(qloc)"])
    plt.ylabel("value")
    plt.title("qloc components")
    plt.show()

    return qloc

if __name__ == "__main__":
    qloc = main()


# ----------------------------
# Physically meaningful qloc plot (complex plane)
# ----------------------------

plt.figure()

# Estimated location from harmonics
plt.scatter(np.real(qloc), np.imag(qloc),
            s=120, marker="o", label="Estimated qloc")

# True source points (field generators)
plt.scatter(np.real(pointsZi), np.imag(pointsZi),
            s=80, marker="s", label="True source points")

# Axes lines for reference
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.axis("equal")
plt.xlabel("Re (x position)")
plt.ylabel("Im (y position)")
plt.title("Estimated disturbance location (qloc) vs true sources")
plt.legend()
plt.grid(True)

plt.show()
```
